```python
import pandas as pd
df=pd.read_csv("dtypes.csv")
```

```python
df
```

```python
df['gender']=df['gender'].map({'m':'M','male':'M','f':'F','female':'F'})
df
```

```python
df['total2']=df[['mark_1','mark_2']].apply(lambda x:x['mark_1']+x['mark_2'],axis=1)
df
```

```python
df['total']=df[['mark_1','mark_2']].apply(lambda x:x.sum(),axis=1)
df

```

```python
df.drop(columns=['total'],inplace=True)
df
```

```python
df['total']=df['mark_1']+df['mark_2']
df
```

```python
def initials(name):
    arr=name.split(' ')
    init=''
   
    init=[l[0] for l in arr]
    return ' '.join(init)
```

```python
df['initials']=df['name'].apply(initials)
df
```

```python
df.drop(columns=['initials'],inplace=True)
df
```

```python
df['ini']=df['name'].transform(initials)
df
```

```python
df['dummy']='';
df
```

```python
def t(row):
    row['percent']=(row['mark_1']+row['mark_2'])/0.4
    return row
df.transform(t,axis=1)
df
```

```python
df['p']=df.transform(lambda row:row['mark_1']+row['mark_2'],axis=1)
df
```

```python
def tt(row):
   row['mytot']= row['mark_1']+row['mark_2']
   return row
df.transform(tt)
df
```
